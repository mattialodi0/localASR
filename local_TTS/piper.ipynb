# Piper TTS

```python
from io import BytesIO
from pathlib import Path
import wave

from piper import PiperVoice

MODEL_PATH = Path("voices/it_IT-paola-medium.onnx")
if not MODEL_PATH.exists():
    raise RuntimeError(
        f"Piper model not found: {MODEL_PATH}. "
        "Download the voice model first."
    )

# Load the model once at startup, not on every request.
voice = PiperVoice.load(str(MODEL_PATH))

def text_to_speech(text: str):
    text = text.strip()


    output = BytesIO()
    try:
        with wave.open(output, "wb") as wav_file:
            voice.synthesize_wav(text, wav_file)
    except Exception as exc:
        raise RuntimeError(f"Failed to synthesize speech: {exc}") from exc
    return output.getvalue()

# run tts
text = "Lorem ipsum dolor sit amet consectetur, adipisicing elit. Sapiente minus praesentium repellendus adipisci nobis quo officia! Quod earum vel itaque soluta cupiditate, repellat voluptatem voluptatibus, magnam repellendus consequuntur animi omnis enim sapiente, obcaecati error! Voluptates est incidunt in sequi quisquam necessitatibus sint doloribus hic numquam saepe, voluptate architecto dolor aperiam culpa cumque placeat ipsum cum, nihil fugiat vero tenetur, molestias sunt. Nostrum cum ad architecto? Neque totam in, expedita illum velit qui. Recusandae harum consequuntur laboriosam! Enim ea nesciunt corporis officiis architecto cum voluptas nobis maxime dolore quam, vitae quia illo perspiciatis sint consequatur ratione dolor similique! Atque, cupiditate quo."
audio_data = text_to_speech(text)
with open("./test_files/output.wav", "wb") as f:
    f.write(audio_data)
print("Audio saved")
```
